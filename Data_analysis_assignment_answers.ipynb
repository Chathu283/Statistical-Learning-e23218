## 1. Bias of the Maximum Likelihood Estimator (MLE) of the Covariance Matrix

The Maximum Likelihood Estimator (MLE) of the covariance matrix is defined as

$$
\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}
=
\frac{1}{n}
\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T .
$$

To determine whether this estimator is unbiased, consider the identity

$$
\sum_{i=1}^{n}
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)^T
=
\sum_{i=1}^{n}
(\mathbf X_i-\boldsymbol{\mu})
(\mathbf X_i-\boldsymbol{\mu})^T
-
n(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})^T .
$$

Taking expectations on both sides gives

$$
\mathbb E
\left[
\sum_{i=1}^{n}
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)^T
\right]
=
\mathbb E
\left[
\sum_{i=1}^{n}
(\mathbf X_i-\boldsymbol{\mu})
(\mathbf X_i-\boldsymbol{\mu})^T
\right]
-
n\,
\mathbb E
\left[
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})^T
\right].
$$

Since the observations are i.i.d.,

$$
\mathbb E
\left[
(\mathbf X_i-\boldsymbol{\mu})
(\mathbf X_i-\boldsymbol{\mu})^T
\right]
=
\boldsymbol{\Sigma},
$$

therefore

$$
\mathbb E
\left[
\sum_{i=1}^{n}
(\mathbf X_i-\boldsymbol{\mu})
(\mathbf X_i-\boldsymbol{\mu})^T
\right]
=
n\boldsymbol{\Sigma}.
$$

Also,

$$
\mathrm{Var}(\widehat{\boldsymbol{\mu}}_n)
=
\frac{\boldsymbol{\Sigma}}{n},
$$

which implies

$$
\mathbb E
\left[
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})^T
\right]
=
\frac{\boldsymbol{\Sigma}}{n}.
$$

Substituting,

$$
\mathbb E
\left[
\sum_{i=1}^{n}
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)^T
\right]
=
n\boldsymbol{\Sigma}
-
n\left(\frac{\boldsymbol{\Sigma}}{n}\right)
=
(n-1)\boldsymbol{\Sigma}.
$$

Therefore,

$$
\mathbb E[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]
=
\frac{1}{n}(n-1)\boldsymbol{\Sigma}.
$$

Hence,

$$
\boxed{
\mathbb E[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]
=
\frac{n-1}{n}\boldsymbol{\Sigma}
}
$$

which proves that the MLE systematically underestimates the true covariance matrix and is therefore biased for finite sample sizes.

---

## Bessel's Correction and the Unbiased Estimator

To remove the bias, the covariance estimator is scaled by the factor

$$
\frac{n}{n-1}.
$$

This yields the sample covariance matrix

$$
\mathbf S
=
\frac{1}{n-1}
\sum_{i=1}^{n}
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)^T .
$$

Taking expectations,

$$
\mathbb E[\mathbf S]
=
\frac{1}{n-1}
\mathbb E
\left[
\sum_{i=1}^{n}
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)^T
\right].
$$

Using the previous result,

$$
\mathbb E[\mathbf S]
=
\frac{1}{n-1}
(n-1)\boldsymbol{\Sigma}
=
\boldsymbol{\Sigma}.
$$

Therefore,

$$
\boxed{
\mathbb E[\mathbf S]
=
\boldsymbol{\Sigma}
}
$$

showing that $\mathbf S$ is unbiased.

Bessel's correction compensates for the loss of one degree of freedom caused by estimating the unknown population mean using the sample mean. Since the correction factor $\frac{n}{n-1}$ exactly removes the bias, $\mathbf S$ is the unique unbiased estimator obtained from the MLE form.

A Type I Error occurs when the monitoring system concludes that structural damage exists even though the structure is actually healthy.

$$
\boxed{
\alpha
=
P(\text{Reject }H_0 \mid H_0 \text{ is true})
}
$$

In practice, this corresponds to a false alarm, which may trigger unnecessary inspections, maintenance actions, or plant shutdowns.

### Type II Error ($\beta$)

A Type II Error occurs when the monitoring system fails to detect actual structural damage.

$$
\boxed{
\beta
=
P(\text{Fail to reject }H_0 \mid H_1 \text{ is true})
}
$$

In practice, this corresponds to a missed detection, allowing a damaged structure to continue operating.

The statistical power of a hypothesis test is defined as

$$
\boxed{
\text{Power}
=

}
$$

## Slutsky's Theorem

Let $\{X_n\}$ and $\{Y_n\}$ be sequences of random variables such that

$$
X_n \xrightarrow{d} X
$$

and

$$
Y_n \xrightarrow{p} c,
$$

where $c$ is a constant.

Then Slutsky's Theorem states that

$$
X_n + Y_n \xrightarrow{d} X + c,
$$

$$
X_n Y_n \xrightarrow{d} cX,
$$

and, provided that $c \neq 0$,

$$
\frac{X_n}{Y_n}
\xrightarrow{d}
\frac{X}{c}.
$$

Thus, any random quantity that converges in probability to a constant may be asymptotically replaced by that constant when combined with another quantity that converges in distribution.

---

## Application to the Sample Covariance Matrix

The Multivariate Central Limit Theorem states that

$$
\sqrt{n}
\left(
\widehat{\boldsymbol{\mu}}_n
-
\boldsymbol{\mu}
\right)
\xrightarrow{d}
\mathscr N
\left(
\mathbf 0,
\boldsymbol{\Sigma}
\right).
$$

However, the true covariance matrix $\boldsymbol{\Sigma}$ is unknown and cannot be observed directly.

Instead, we estimate it using the sample covariance matrix

$$
\mathbf S
=
\frac{1}{n-1}
\sum_{i=1}^{n}
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf X_i-\widehat{\boldsymbol{\mu}}_n)^T.
$$

By the Law of Large Numbers, the sample covariance matrix is a consistent estimator of the population covariance matrix:

$$
\mathbf S
\xrightarrow{p}
\boldsymbol{\Sigma}.
$$

Since matrix inversion and square-root operations are continuous transformations for positive-definite matrices, it follows that

$$
\mathbf S^{-1/2}
\xrightarrow{p}
\boldsymbol{\Sigma}^{-1/2}.
$$

Multiplying the Multivariate CLT by $\mathbf S^{-1/2}$ gives

$$
\mathbf S^{-1/2}
\sqrt n
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu}).
$$

Applying Slutsky's Theorem,

$$
\mathbf S^{-1/2}
\sqrt n
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
\xrightarrow{d}
\boldsymbol{\Sigma}^{-1/2}
Z,
$$

where

$$
Z
\sim
\mathscr N(\mathbf 0,\boldsymbol{\Sigma}).
$$

Since

$$
\boldsymbol{\Sigma}^{-1/2}Z
\sim
\mathscr N(\mathbf 0,\mathbf I),
$$

we obtain

$$
\mathbf S^{-1/2}
\sqrt n
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
\xrightarrow{d}
\mathscr N(\mathbf 0,\mathbf I).
$$

Therefore, replacing the unknown covariance matrix $\boldsymbol{\Sigma}$ with its consistent estimator $\mathbf S$ does not alter the asymptotic distribution.

---

## Practical Parametric Modeling Distribution

The Multivariate CLT implies that for sufficiently large sample sizes,

$$
\widehat{\boldsymbol{\mu}}_n
\approx
\mathscr N
\left(
\boldsymbol{\mu},
\frac{1}{n}\boldsymbol{\Sigma}
\right).
$$

Since

$$
\mathbf S
\xrightarrow{p}
\boldsymbol{\Sigma},
$$

Slutsky's Theorem justifies replacing $\boldsymbol{\Sigma}$ by $\mathbf S$, yielding the operational distribution

$$
\widehat{\boldsymbol{\mu}}_n
\approx
\mathscr N
\left(
\boldsymbol{\mu},
\frac{1}{n}\mathbf S
\right).
$$

In practice, the unknown mean $\boldsymbol{\mu}$ is also replaced by its empirical estimate $\widehat{\boldsymbol{\mu}}_n$, producing the fully data-driven approximation used in the dashboard:

$$
\boxed{
\widehat{\boldsymbol{\mu}}_n
\sim
\mathscr N
\left(
\widehat{\boldsymbol{\mu}}_n,
\frac{1}{n}\mathbf S
\right)
}
$$

Thus, Slutsky's Theorem provides the theoretical justification for substituting the latent population covariance matrix with the empirical sample covariance matrix when constructing confidence ellipsoids, monitoring statistics, and multivariate diagnostic models for large sample sizes.

***Scenario:** You are provided an industrial dataset containing $n = 5,000$ snapshots across $m = 3$ sensor channels measuring physical strain. To verify the weak i.i.d. assumption before initializing a global PCA or Factor Analysis layer, you must explicitly test **Condition 1: Constant Joint Mean (First Moment Homogeneity)** across the time index.*

**Your Task:**
Write a clean, self-contained Python function within Google Colab to partition the dataset into $g = 5$ consecutive, non-overlapping blocks (chunks) of size $n_j = 1,000$. Implement a Multivariate Analysis of Variance (MANOVA) tracking framework from scratch (or leveraging `statsmodels`) to calculate **Wilks' Lambda ($\Lambda$)**.

$$\Lambda = \frac{|\mathbf{W}|}{|\mathbf{B} + \mathbf{W}|}$$

Transform your calculated $\Lambda$ using **Bartlett’s $\chi^2$ approximation**:


$$\chi^2_{\text{calc}} = -\left(n - 1 - \frac{m + g}{2}\right) \ln \Lambda$$

Execute your code using the synthetic data generator provided below. Report your exact calculated Wilks' $\Lambda$, the Bartlett $\chi^2$ test statistic, and its corresponding $p$-value under $\chi^2 \sim \chi^2_{m(g-1)}$. Based on an alpha threshold of $\alpha = 0.05$, state your final diagnostic conclusion: Has the baseline center shifted over time, or is the first moment homogeneous?



````python
import numpy as np
import pandas as pd
import scipy.stats as stats

# --- DO NOT MODIFY: Synthetic Strain Sensor Data Generator ---
np.random.seed(42)
n_samples = 5000
n_features = 3

# Generating base stationary multivariate normal noise
base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

# Injecting a subtle, hidden structural drift in the final 1,000 snapshots
base_data[4000:, 0] += 0.015  # Sensor 1 drift
base_data[4000:, 2] -= 0.010  # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ----------------------------------------------------------------

def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g chunks and evaluates first-moment homogeneity
    via Wilks' Lambda and Bartlett's Chi-Square asymptotic transformation.
    """
    # Convert dataframe values to raw NumPy arrays for high-performance linear algebra
    X = df.values
    n, m = X.shape
    n_j = n // g_chunks  # Samples per chunk (1,000)

    # 1. Compute the global overall grand mean vector
    grand_mean = np.mean(X, axis=0)

    # Initialize error partitions
    W = np.zeros((m, m))
    B = np.zeros((m, m))

    # 2. Iterate through temporal chunks to partition variance
    for j in range(g_chunks):
        # Extract consecutive non-overlapping block
        chunk = X[j * n_j : (j + 1) * n_j]
        chunk_mean = np.mean(chunk, axis=0)

        # Within-Chunk Error Sum of Squares & Cross-Products
        dev_within = chunk - chunk_mean
        W += dev_within.T @ dev_within

        # Between-Chunk Hypothesis Sum of Squares & Cross-Products
        dev_between = (chunk_mean - grand_mean).reshape(-1, 1)
        B += n_j * (dev_between @ dev_between.T)

    # 3. Compute Wilks' Lambda (Determinant Ratio)
    det_W = np.linalg.det(W)
    det_Total = np.linalg.det(B + W)
    wilks_lambda = det_W / det_Total

    # 4. Apply Bartlett's Chi-Square Asymptotic Transformation
    df_multiplier = n - 1 - (m + g_chunks) / 2.0
    chi2_calc = -df_multiplier * np.log(wilks_lambda)

    # Degrees of freedom calculation: m * (g - 1)
    df_chi2 = m * (g_chunks - 1)

    # Compute asymptotic p-value using survival function
    p_value = stats.chi2.sf(chi2_calc, df_chi2)

    # Pack analytical outputs into a dictionary format
    results = {
        "Wilks_Lambda": float(wilks_lambda),
        "Chi2_Stat": float(chi2_calc),
        "Degrees_of_Freedom": int(df_chi2),
        "p_value": float(p_value),
        "Is_Homogeneous": bool(p_value >= 0.05)
    }

    return results

# Execute your completed function below:
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)

# Clear printing format
print("--- Homogeneity Analysis Output ---")
for key, val in results.items():
    print(f"{key:<20}: {val}")
````

\boldsymbol{\Sigma}

\mathbb{E}

````python
import numpy as np

class StructuralDiagnostics:
    def __init__(self, baseline_data: np.ndarray):
        """
        Parameters:
        -----------
        baseline_data : np.ndarray
            A 2D array of shape (n, m) representing 'n' nominal snapshots
            across 'm' sensors.
        """
        self.X = baseline_data
        self.n, self.m = baseline_data.shape

    def analyze_subspace_boundaries(self) -> dict:
        """
        Analyzes Hotelling's T^2 and Q statistic empirical means
        for truncation boundaries k in {1, 2, 3}.

        Returns:
        --------
        dict
            A dictionary mapped by k containing the empirical means of T^2 and Q.
        """
        # 1. Standardize and center the incoming realizations
        # Compute mean and standard deviation for each sensor (column-wise)
        means = np.mean(self.X, axis=0)
        stds = np.std(self.X, axis=0, ddof=1)  # Using sample std dev

        # Avoid division by zero if a sensor has constant values
        stds[stds == 0] = 1.0

        # Center and scale
        Z = (self.X - means) / stds

        # 2. Decompose the unbiased sample covariance matrix S
        # Since Z is standardized, S is technically the sample correlation matrix
        S = np.cov(Z, rowvar=False, ddof=1)

        # np.linalg.eigh returns eigenvalues in ascending order
        eigenvalues, eigenvectors = np.linalg.eigh(S)

        # Sort components in descending order
        idx = np.argsort(eigenvalues)[::-1]
        lambdas = eigenvalues[idx]
        V = eigenvectors[:, idx]

        # Transform standardized data into the Principal Component (score) space
        # Z_proj shape will be (n, m) where columns represent score vectors z_j
        Z_proj = Z @ V

        # Dictionary to store the resulting empirical means
        results = {}

        # 3. Loop through every possible subspace cutoff value k
        for k in range(1, self.m):  # m = 4, so k loops through 1, 2, 3

            # Hotelling's T^2 calculation for each sample i:
            # T^2_i = \sum_{j=1}^k (z_{i,j}^2 / \lambda_j)
            T2_samples = np.sum((Z_proj[:, :k] ** 2) / lambdas[:k], axis=1)

            # Q statistic calculation for each sample i:
            # Q_i = \sum_{j=k+1}^m z_{i,j}^2
            Q_samples = np.sum(Z_proj[:, k:] ** 2, axis=1)

            # 4. Compute the expected value (empirical mean) across all samples
            mean_T2 = np.mean(T2_samples)
            mean_Q = np.mean(Q_samples)

            # 5. Store metrics
            results[k] = {
                "empirical_mean_T2": mean_T2,
                "empirical_mean_Q": mean_Q
            }

        return results

# --- Verification Block (Optional) ---
# If you want to test the implementation directly in your Colab cell:
if __name__ == "__main__":
    # Generate synthetic nominal data: n = 3000 snapshots, m = 4 sensors
    np.random.seed(42)
    synthetic_data = np.random.normal(loc=10.0, scale=2.5, size=(3000, 4))

    # Run pipeline
    diagnostics = StructuralDiagnostics(synthetic_data)
    boundary_metrics = diagnostics.analyze_subspace_boundaries()

    # Print results to verify against your Part A theoretical derivations
    print(f"{'k':<5} | {'E[T^2] (Empirical)':<20} | {'E[Q] (Empirical)':<20}")
    print("-" * 52)
    for k, metrics in boundary_metrics.items():
        print(f"{k:<5} | {metrics['empirical_mean_T2']:<20.4f} | {metrics['empirical_mean_Q']:<20.4f}")
````

## **Part A: Theoretical Fundamentals (Generative Model & Commonalities)**

$$\text{ 1. The Fundamental Equation and Variance Decomposition}$$

$$\text{ Proof of the Fundamental Equation of Factor Analysis}$$

$$\text{Given the linear generative model for a normalized, zero-mean baseline feature vector } \mathbf{Z}_i \in \mathbb{R}^m:$$

$$\mathbf{Z}_i = \boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i$$

$$\text{The population covariance matrix (which matches the universal correlation matrix } \mathbf{R} \text{ because } \mathbf{Z}_i \text{ is standardized) is defined by the expectation:}$$

$$\mathbf{R} = \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T]$$

$$\text{Substituting the structural equation into the expectation expression:}$$

$$\mathbf{R} = \mathbb{E}\left[ (\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)(\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)^T \right]$$

$$\text{Applying the matrix transpose rule } (\mathbf{A} + \mathbf{B})^T = \mathbf{A}^T + \mathbf{B}^T:$$

$$\mathbf{R} = \mathbb{E}\left[ (\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)(\mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\epsilon}_i^T) \right]$$

$$\text{Expanding the terms inside using the linearity of expectation:}$$

$$\mathbf{R} = \mathbb{E}[\boldsymbol{\Lambda} \mathbf{F}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T] + \mathbb{E}[\boldsymbol{\Lambda} \mathbf{F}_i \boldsymbol{\epsilon}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T] + \mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T]$$

$$\text{Since the factor loading matrix } \boldsymbol{\Lambda} \text{ contains deterministic, constant parameters, it can be factored outside of the expectation operators:}$$

$$\mathbf{R} = \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T]$$

$$\text{We now substitute the core statistical assumptions governing the structural model:}$$

$$\text{1. Orthogonal common factors: } \mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] = \mathbf{I}_{k \times k}$$

$$\text{2. Statistically independent factors and errors: } \mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] = \mathbf{0}_{m \times k} \text{ and } \mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] = \mathbf{0}_{k \times m}$$

$$\text{3. Diagonalized measurement unique variance: } \mathbb{E}[\boldsymbol{\epsilon}_i\boldsymbol{\epsilon}_i^T] = \boldsymbol{\Psi}$$

$$\text{Evaluating the terms yields:}$$

$$\mathbf{R} = \boldsymbol{\Lambda} (\mathbf{I}_{k \times k}) \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} (\mathbf{0}_{k \times m}) + (\mathbf{0}_{m \times k}) \boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$

$$\mathbf{R} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$

$$\text{---}$$

$$\text{ Diagonal Decomposition \& Diagnostics Interpretation}$$

$$\text{Evaluating the } j\text{-th diagonal element of the fundamental relation } \mathbf{R} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}\text{, the total unit variance of the } j\text{-th standardized sensor } (R_{jj} = 1) \text{ decomposes into:}$$

$$1 = \sum_{l=1}^k \lambda_{jl}^2 + \varphi_j^2$$

$$\text{ Communality } (h_j^2)$$

$$\text{Definition: } h_j^2 = \sum_{l=1}^k \lambda_{jl}^2$$

$$\text{* Physical Interpretation: This represents the shared portion of variance for sensor } j \text{ driven by global, underlying structural or operational phenomena (e.g., ambient temperature cycles, engine RPM shifts, system-wide load distributions). In asset diagnostics, a high communality means the sensor behaves predictably within the global network.}$$

$$\text{Uniqueness } (\varphi_j^2)$$

$$\text{Definition: } \varphi_j^2 = 1 - h_j^2$$

$$\text{* Physical Interpretation: This encompasses all variance completely unique to that specific node. It comprises local structural variations (such as localized structural fatigue or localized thermal gradients) plus raw measurement noise.}$$

$$\text{* Diagnostics Application: Sudden shifts in } \varphi_j^2 \text{ signal localized component anomalies or an isolated sensor calibration failure, preventing local tracking errors from corrupting global state estimations.}$$

$$\text{---}$$

$$\text{ 2. Factor Rotation and Interpretability}$$

$$\text{Principal Component Loadings vs. Varimax-Rotated Loadings}$$

$$\text{* **Raw PCA Loadings:** Maximize variance sequentially along orthogonal axes. This yields a dense loading matrix where nearly every sensor exhibits moderate to high weights on the first major components, making root-cause isolation highly coupled and physically ambiguous.}$$

$$\text{* **Varimax-Rotated Loadings:** Performs an orthogonal rotation to maximize the variance of the squared loadings within each column. This drives the loading matrix to a **simple structure**, meaning each column contains elements near } \pm 1 \text{ or } 0\text{. Physically, it uncouples the components, isolating distinct structural subsystems so that an anomaly pinpointed on a factor maps directly to a discrete, physical cluster of sensors.}$$

$$\text{---}$$

$$\text{ Invariance of Communality and Global Covariance Under Orthogonal Rotations}$$

$$\text{Let } \mathbf{T} \text{ be an orthogonal } k \times k \text{ transformation matrix satisfying } \mathbf{T}\mathbf{T}^T = \mathbf{T}^T\mathbf{T} = \mathbf{I}_{k \times k}\text{. The rotated loading matrix is defined as } \boldsymbol{\Lambda}^* = \boldsymbol{\Lambda}\mathbf{T}\text{.}$$

$$\text{Global Covariance Invariance}$$

$$\text{Reconstructing the correlation matrix using the rotated loadings yields:}$$

$$\mathbf{R}^* = \boldsymbol{\Lambda}^*(\boldsymbol{\Lambda}^*)^T + \boldsymbol{\Psi}$$

$$\mathbf{R}^* = (\boldsymbol{\Lambda}\mathbf{T})(\boldsymbol{\Lambda}\mathbf{T})^T + \boldsymbol{\Psi}$$

$$\text{Using the matrix transpose identity } (\mathbf{A}\mathbf{B})^T = \mathbf{B}^T\mathbf{A}^T:$$

$$\mathbf{R}^* = \boldsymbol{\Lambda}\mathbf{T}\mathbf{T}^T\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$

$$\text{Substituting } \mathbf{T}\mathbf{T}^T = \mathbf{I}_{k \times k} \text{ shows that the rotation cancels out internally, preserving the exact global approximation:}$$

$$\mathbf{R}^* = \boldsymbol{\Lambda}\mathbf{I}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi} = \mathbf{R}$$

$$\text{Communality Invariance}$$

$$\text{Let } \boldsymbol{\lambda}_j^T \text{ be the } j\text{-th row vector of } \boldsymbol{\Lambda}\text{. The corresponding row vector of the rotated matrix is given by } (\boldsymbol{\lambda}_j^*)^T = \boldsymbol{\lambda}_j^T\mathbf{T}\text{.}$$

$$\text{The rotated communality } (h_j^*)^2 \text{ is the squared Euclidean norm of this transformed row vector:}$$

$$(h_j^*)^2 = \|\boldsymbol{\lambda}_j^*\|_2^2 = (\boldsymbol{\lambda}_j^*)^T \boldsymbol{\lambda}_j^*$$

$$\text{Substituting the transformed row expression:}$$

$$(h_j^*)^2 = (\boldsymbol{\lambda}_j^T\mathbf{T})(\boldsymbol{\lambda}_j^T\mathbf{T})^T = \boldsymbol{\lambda}_j^T\mathbf{T}\mathbf{T}^T\boldsymbol{\lambda}_j$$

$$\text{Since } \mathbf{T}\mathbf{T}^T = \mathbf{I}_{k \times k}:$$

$$(h_j^*)^2 = \boldsymbol{\lambda}_j^T\mathbf{I}\boldsymbol{\lambda}_j = \boldsymbol{\lambda}_j^T\boldsymbol{\lambda}_j = \|\boldsymbol{\lambda}_j\|_2^2 = h_j^2$$

$$\text{This proves that rigid orthogonal rotations simply pivot the coordinate axes within the latent feature subspace. While this shifts individual column metrics to simplify physical interpretation, it leaves both individual sensor communalities } (h_j^2) \text{ and the reconstructed global correlation matrix } (\mathbf{R}) \text{ unchanged.}$$

1. Z-Score Mapping leyer



````python
import numpy as np
import pandas as pd
from scipy.linalg import svd
from sklearn.decomposition import FactorAnalysis

EPS = 1e-15


def zscore_layer(df: pd.DataFrame):
    X = df.select_dtypes(include=[np.number]).to_numpy(dtype=float)

    mu = np.mean(X, axis=0)
    sigma = np.std(X, axis=0, ddof=1)

    sigma = np.where(sigma < EPS, EPS, sigma)

    Z = (X - mu) / sigma

    return Z, mu, sigma
````

2. Dimensionality Constraint Layer

````python
def validate_dimensions(Z, k):
    n, m = Z.shape

    # Rule 1: sample size must exceed feature count
    if n <= m:
        raise ValueError(
            f"Invalid dataset: sample size n={n} must be greater than feature count m={m}."
        )

    # Rule 2: latent dimension constraint
    if k >= m:
        raise ValueError(
            f"Invalid latent dimension: k={k} must be strictly less than feature dimension m={m}."
        )

    return n, m
````

3. Varimax Rotation (Orthogonal Rotation Matrix T)

````python
def varimax(Phi, gamma=1.0, q=50, tol=1e-6):
    p, k = Phi.shape
    R = np.eye(k)
    d = 0

    for _ in range(q):
        d_old = d
        Lambda = np.dot(Phi, R)

        u, s, v = svd(
            Phi.T @ (Lambda**3 - (gamma / p) * Lambda @ np.diag(np.diag(Lambda.T @ Lambda)))
        )

        R = u @ v
        d = np.sum(s)

        if d_old != 0 and d - d_old < tol:
            break

    return Phi @ R
````

4. Information Decomposition Layer (Factor Analysis)

````python
def factor_decomposition(Z, k):
    n, m = Z.shape

    # Fit Factor Analysis model (Maximum Likelihood)
    fa = FactorAnalysis(n_components=k, random_state=0)
    fa.fit(Z)

    # Factor loadings
    loadings = fa.components_.T  # shape: (m, k)

    # Apply Varimax rotation
    rotated_loadings = varimax(loadings)

    # Communality: sum of squared loadings per feature
    h2 = np.sum(rotated_loadings**2, axis=1)

    # Uniqueness: residual variance per feature
    phi2 = np.maximum(1 - h2, 0)

    return {
        "loadings": rotated_loadings,
        "communality": h2,
        "uniqueness": phi2
    }
````

5. Full Pipeline Wrapper

````python
def empirical_decomposition(df, k):
    # Step 1: Standardization
    Z, mean, std = zscore_standardize(df)

    # Step 2: Validation
    n, m = validate_dimensions(Z, k)

    # Step 3–4: Factor model + decomposition
    results = factor_decomposition(Z, k)

    return {
        "Z": Z,
        "mean": mean,
        "std": std,
        "n": n,
        "m": m,
        **results
    }
````

````python
import numpy as np
import pandas as pd
from typing import Optional, Sequence, Dict, Any
from sklearn.decomposition import FactorAnalysis
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# =====================================================================
# COMBINED SUBSPACE ENGINES (PCA & FA PIPELINES)
# =====================================================================

class DualSubspaceDiagnosticsEngine:
    def __init__(self, df: pd.DataFrame):
        self.df = df

    def compute_empirical_pca(self, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the geometric de-correlation framework via PCA.
        Decomposes the unbiased empirical covariance matrix S into its orthogonal
        basis P_hat and renders a clean 2x3 optimization and feature contribution dashboard.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        if n <= m:
            raise ValueError("Sample size n must be larger than feature dimensions m to isolate principal components.")

        # 1. Centering
        mu_hat = np.mean(X, axis=0)
        X_centered = X - mu_hat

        # 2. Compute Unbiased Sample Covariance Matrix S
        S_matrix = np.cov(X, rowvar=False, ddof=1)

        # 3. Spectral Decomposition
        eigenvalues, eigenvectors = np.linalg.eigh(S_matrix)
        idx = np.argsort(eigenvalues)[::-1]
        lambda_hat = eigenvalues[idx]
        P_hat = eigenvectors[:, idx]
        lambda_hat = np.clip(lambda_hat, a_min=1e-15, a_max=None)

        # 4. Variance Information Calculations
        total_variance = np.sum(lambda_hat)
        if total_variance == 0:
            raise ValueError("Total variance is zero.")

        explained_variance_ratio = lambda_hat / total_variance
        cumulative_variance_ratio = np.cumsum(explained_variance_ratio)
        unexplained_variance_ratio = 1.0 - cumulative_variance_ratio

        # 5. Map Realizations to Principal Component Scores
        Z_scores = np.dot(X_centered, P_hat)
        S_Z = np.cov(Z_scores, rowvar=False, ddof=1)

        # 6. Evaluate Statistical Distance Metrics (T^2 and Q)
        k_range = np.arange(1, m)
        mean_T2_vs_k = []
        mean_Q_vs_k = []
        T2_matrix = np.zeros((n, len(k_range)))
        Q_matrix = np.zeros((n, len(k_range)))

        for idx_k, k in enumerate(k_range):
            Z_k = Z_scores[:, :k]
            lambda_k = lambda_hat[:k]
            T2_samples = np.sum((Z_k ** 2) / lambda_k, axis=1)
            T2_matrix[:, idx_k] = T2_samples
            mean_T2_vs_k.append(np.mean(T2_samples))

            Z_residual = Z_scores[:, k:]
            Q_samples = np.sum(Z_residual ** 2, axis=1)
            Q_matrix[:, idx_k] = Q_samples
            mean_Q_vs_k.append(np.mean(Q_samples))

        print(f"\n--- Operationalizing Geometric De-correlation Layer (PCA) ---")
        print(f"Decomposing structural space of {m} features using {n} samples.")
        print(f"Total System Variance (Trace[S]): {total_variance:.4f}")

        # 7. Enhanced 2x3 Diagnostic Dashboard
        if show_plot:
            pc_labels = [f"PC {i+1}" for i in range(m)]
            k_labels = [f"k={k}" for k in k_range]
            sensor_labels = target_cols

            fig = make_subplots(
                rows=2, cols=3,
                horizontal_spacing=0.18,
                vertical_spacing=0.28,
                subplot_titles=(
                    "Feature Loading Matrix |P_hat|",
                    "Component Values (Eigenvalues λ)",
                    "Information Profile (Explained Var.)",
                    "Residual Space (Unexplained Var.)",
                    "Mean Hotelling's T² vs Subspace Size k",
                    "Mean Q Statistic (SPE) vs Subspace Size k"
                )
            )

            # --- ROW 1, COL 1: PCA Loading Heatmap (Left-anchored Colorbar) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(P_hat),
                    x=pc_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Loading Weight",
                        x=-0.12,          # Perfectly isolates colorbar to the left side
                        len=0.38,         # Fits height bounds of row 1
                        y=0.78,           # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right",
                        titleside="top"
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Principal Axes", row=1, col=1)

            # --- ROW 1, COL 2: Absolute Eigenvalues ---
            fig.add_trace(go.Bar(x=pc_labels, y=lambda_hat, name="Eigenvalue (λ_j)", marker=dict(color='#1f77b4', line=dict(color='black', width=0.5)), legendgroup="eigen"), row=1, col=2)
            fig.update_yaxes(title_text="Variance Magnitude", row=1, col=2)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=2)

            # --- ROW 1, COL 3: Variance Profiles ---
            fig.add_trace(go.Bar(x=pc_labels, y=explained_variance_ratio * 100, name="Marginal Explained", marker=dict(color='#ff7f0e', opacity=0.75), legendgroup="expl"), row=1, col=3)
            fig.add_trace(go.Scatter(x=pc_labels, y=cumulative_variance_ratio * 100, mode='lines+markers', name='Cumulative Captured', line=dict(color='#d62728', width=2.5, dash='dash'), legendgroup="expl"), row=1, col=3)
            fig.update_yaxes(title_text="Captured Structure (%)", range=[-2, 105], row=1, col=3)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=3)

            # --- ROW 2, COL 1: Remaining Noise ---
            fig.add_trace(go.Bar(x=pc_labels, y=unexplained_variance_ratio * 100, name="Remaining Noise", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="noise"), row=2, col=1)
            fig.update_yaxes(title_text="Excluded Info (%)", range=[-2, 105], row=2, col=1)
            fig.update_xaxes(title_text="Principal Axes", row=2, col=1)

            # --- ROW 2, COL 2: Hotelling's T^2 ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_T2_vs_k, mode='lines+markers', name='Mean T²', line=dict(color='#9467bd', width=2.5), marker=dict(size=6, symbol='diamond'), legendgroup="t2"), row=2, col=2)
            fig.update_yaxes(title_text="Average T² Metric", row=2, col=2)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=2)

            # --- ROW 2, COL 3: SPE / Q Statistic ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_Q_vs_k, mode='lines+markers', name='Mean Q (SPE)', line=dict(color='#e377c2', width=2.5), marker=dict(size=6, symbol='square'), legendgroup="q_stat"), row=2, col=3)
            fig.update_yaxes(title_text="Average Residual Energy", row=2, col=3)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=3)

            # Global Unified Layout Configurations
            fig.update_layout(
                title=dict(text="Principal Component Analysis (PCA) Optimization & Feature Loading Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5
                ),
                margin=dict(t=150, b=60, l=140, r=80),
                height=750,
                width=1250
            )
            fig.show()

        return {
            "mean_vector": mu_hat, "covariance_matrix_S": S_matrix, "eigenvalues_lambda": lambda_hat,
            "eigenvectors_P": P_hat, "explained_variance_ratio": explained_variance_ratio,
            "cumulative_variance_ratio": cumulative_variance_ratio, "unexplained_variance_ratio": unexplained_variance_ratio,
            "transformed_scores_Z": Z_scores, "score_covariance_diagonal": np.diag(S_Z),
            "features": target_cols, "k_values": k_range, "T2_matrix_vs_k": T2_matrix, "Q_matrix_vs_k": Q_matrix,
            "mean_T2_profile": np.array(mean_T2_vs_k), "mean_Q_profile": np.array(mean_Q_vs_k)
        }

    def compute_empirical_fa(self, k: int, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the Factor Analysis latent subspace framework with a clean 2x2 grid.
        Anchors the heatmap colorbar perfectly to the right edge of the first subplot.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        # 1. Standardize realizations to construct Z space
        mu_hat = np.mean(X, axis=0)
        std_hat = np.std(X, axis=0, ddof=1)
        std_hat[std_hat == 0] = 1e-15
        Z = (X - mu_hat) / std_hat
        R_matrix = np.corrcoef(X, rowvar=False)

        # 2. Fit Factor Analysis Model (Varimax Rotation)
        fa = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
        fa.fit(Z)
        lambda_matrix = fa.components_.T
        psi_diagonal = fa.noise_variance_

        # 3. Calculate Communality (h²) and Uniqueness (φ²) vectors
        communality = np.sum(lambda_matrix**2, axis=1)
        uniqueness = psi_diagonal

        # 4. Map realizations using Thomson's Regression Method
        F_scores = fa.transform(Z)

        print(f"\n--- Operationalizing Latent Subspace Layer (Factor Analysis) ---")
        print(f"Decomposing structural space of {m} sensors into {k} hidden physical factors.")
        print(f"Average System Communality (Shared Subspace Energy): {np.mean(communality)*100:.2f}%")
        print(f"Average System Uniqueness (Localized Noise Floor): {np.mean(uniqueness)*100:.2f}%")

        # 5. Clean 2x2 Dashboard
        if show_plot:
            sensor_labels = target_cols
            factor_labels = [f"Factor {j+1}" for j in range(k)]

            fig = make_subplots(
                rows=2, cols=2, horizontal_spacing=0.24, vertical_spacing=0.28,
                subplot_titles=(
                    "Structural Loadings Matrix |λ_(j,r)|",
                    "Variance Partitioning (Communality vs Uniqueness)",
                    "Sensor Uniqueness Noise Floor (φ²)",
                    "Latent Factor Scores Empirical Variance"
                )
            )

            # --- Subplot 1: Factor Loadings Heatmap (Colorbar repositioned to x=0.42) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(lambda_matrix),
                    x=factor_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Sensitivity",
                        x=-0.1,          # Shifts colorbar out of the far right to center-left
                        len=0.38,         # Scales down length to match row 1 height bounds
                        y=0.78,          # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right"
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Latent Structures", row=1, col=1)

            # --- Subplot 2: Variance Partitioning ---
            fig.add_trace(go.Bar(y=sensor_labels, x=communality * 100, name="Communality (h² - Shared Structure)", orientation='h', marker=dict(color='#1f77b4'), legendgroup="communality"), row=1, col=2)
            fig.add_trace(go.Bar(y=sensor_labels, x=uniqueness * 100, name="Uniqueness (φ² - Channel Noise)", orientation='h', marker=dict(color='#ff7f0e'), legendgroup="uniqueness_bar"), row=1, col=2)
            fig.update_layout(barmode='stack')
            fig.update_xaxes(title_text="Variance Allocation (%)", range=[0, 100], row=1, col=2)

            # --- Subplot 3: Uniqueness Line Profile ---
            fig.add_trace(go.Scatter(x=sensor_labels, y=uniqueness, mode='lines+markers', name='Uniqueness Profile (φ²)', line=dict(color='#d62728', width=2, dash='dot'), marker=dict(size=8, symbol='x'), legendgroup="uniqueness_line"), row=2, col=1)
            fig.update_yaxes(range=[-0.05, 1.05], row=2, col=1)
            fig.update_xaxes(title_text="Monitored Channels", tickangle=25, row=2, col=1)

            # --- Subplot 4: Latent Factor Scores Empirical Variance ---
            factor_variances = np.var(F_scores, axis=0, ddof=1)
            fig.add_trace(go.Bar(x=factor_labels, y=factor_variances, name="Factor Empirical Variance", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="variance"), row=2, col=2)
            fig.update_yaxes(title_text="Variance Level", row=2, col=2)
            fig.update_xaxes(title_text="Latent Vectors", row=2, col=2)

            # Global Layout Updates
            fig.update_layout(
                title=dict(text="Factor Analysis (FA) Latent Subspace Diagnostics Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5,
                    traceorder="normal",
                    itemwidth=40,
                    itemsizing="constant"
                ),
                margin=dict(t=100, b=60, l=80, r=80),
                height=750,
                width=1100
            )
            fig.show()

        return {
            "mean_vector_mu": mu_hat, "std_vector_D": std_hat, "correlation_matrix_R": R_matrix,
            "factor_loadings_lambda": lambda_matrix, "uniqueness_psi": uniqueness,
            "communality_h2": communality, "latent_factor_scores_F": F_scores, "sensors": target_cols
        }

# =====================================================================
# DATA GENERATOR & PIPELINE EXECUTION
# =====================================================================
if __name__ == "__main__":
    # Set execution seed for perfect reproducible properties
    np.random.seed(42)
    n_samples = 2500

    # 1. Generate underlying, independent true physical driving forces (Latent Factors)
    f1 = np.random.normal(0, 1, n_samples)  # Primary structural mode/load
    f2 = np.random.normal(0, 1, n_samples)  # Secondary operational mode

    # 2. Construct the Observable Multi-Sensor Array with targeted noise profiles
    # Sensors 1 and 2 couple strongly to the primary process (f1)
    s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.3, n_samples)
    s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)

    # Sensor 3 couples strongly to the secondary process (f2)
    s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)

    # Sensor 4 suffers an extreme electrical/instrumentation fault (massive localized white noise)
    s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)

    # 3. Compile into the final evaluation DataFrame
    df_asset = pd.DataFrame(
        data=np.vstack([s1, s2, s3, s4]).T,
        columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
    )
    # Instantiate Engine
    engine = DualSubspaceDiagnosticsEngine(df_asset)

    # Run PCA Engine (1x5 Dashboard)
    pca_metrics = engine.compute_empirical_pca(show_plot=True)

    # Run FA Engine (2x2 Dashboard, Retaining k=2 Common Physical Factors)
    fa_metrics = engine.compute_empirical_fa(k=2, show_plot=True)
````

* **Part 2 (Operational Interpretability):** The PCA loading matrix forces a complex interpretation. ### For Question 4: Operational Trade-offs in System Health Monitoring

* **The PCA Strategy:** Tracking the main principal subspace via $T^2$ alongside the residual space via the $Q$ statistic.
* *Vulnerability:* This approach is highly vulnerable to sensor degradation. As shown in the PCA loading matrix, if a single channel (`Sensor_4`) fails high, it structurally shifts the orientation of your principal axes, dragging the calculations for the remaining healthy channels along with it.


* **The FA Strategy:** Directly monitoring the rotated Latent Factor Scores.
* *Robustness:* This approach is highly robust. Because Factor Analysis explicitly separates localized noise via the diagonal uniqueness vector ($\Psi$), any single-channel sensor failure is safely absorbed by that channel's uniqueness profile ($\psi_4^2 \rightarrow 1$). The underlying latent factors (`Factor 1` and `Factor 2`) remain clean, stable, and uncorrupted, allowing the rest of your health monitoring network to function normally.
